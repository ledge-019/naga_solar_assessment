```python
import os
import sys
import subprocess
import grass.script as gs
import grass.script.setup as gsetup

grass_bin = r"C:\Program Files\QGIS 3.40.13\bin\grass84.bat"

# 1. Get GRASS base
gisbase = subprocess.check_output(
    [grass_bin, "--config", "path"],
    text=True
).strip()

# 2. Get Python bindings path
python_path = subprocess.check_output(
    [grass_bin, "--config", "python_path"],
    text=True
).strip()

# 3. Register GRASS in Python
os.environ["GISBASE"] = gisbase
sys.path.append(python_path)

# 4. CRITICAL: initialize session
gsetup.init(
    gisbase,
    r"C:\Users\YOURNAME\Documents\grassdata",  # <-- change this
    "your_location",                            # <-- change this
    "PERMANENT"
)

# 5. Now GRASS commands work
print(gs.version())
```
